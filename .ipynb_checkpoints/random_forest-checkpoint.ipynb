```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn import metrics
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
df_train = pd.read_csv("wildfires_training.csv")
df_test = pd.read_csv("wildfires_test.csv")
```

```python
#Convert categorical data to numeric
df_train["fire"] = np.where(df_train["fire"]=="no",0,1)
df_test["fire"] = np.where(df_test["fire"]=="no", 0, 1)
```

```python
#Test if yes and no data was converted to 1
df_train["fire"].unique()
```

```python
#Training label
Y = df_train["fire"]

#Training features
X = df_train[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]

#Test label
Y_test = df_test["fire"]

#Test features
X_test = df_test[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]
#Hyperparamter values
trees = [100, 200, 300 , 400, 500]
features = ["sqrt", "log2", None]
results = {f: {"train": [], "test": []} for f in features}

#Function to train and test 
def tuner(n_estimators, max_features):
    #Set hyperparameter values
    model = RandomForestClassifier(
        n_estimators=n_estimators,
        max_features = max_features
    )
    #Fit model on the training data
    model.fit(X, Y)
    #Start predictions
    train_predict=model.predict(X)
    test_predict = model.predict(X_test)
    #Calculate accuracies
    train_accuracy = accuracy_score(Y,train_predict)
    test_accuracy = accuracy_score(Y_test,test_predict)
    print(f"For n_estimators= {n_estimators} max_features={max_features} train accuracy ={train_accuracy} accuracy={test_accuracy}")
    #Store accuracies for results
    results[max_features]["train"].append(train_accuracy)
    results[max_features]["test"].append(test_accuracy)

#Loop through all possible combinations of parameters
for t in trees:
    for f in features:
        tuner(n_estimators=t, max_features=f)
#Plot training and or test data
plt.figure(figsize=(8,5))
for f in features:
    #plt.plot(range(len(trees)), results[f]["train"], marker="o", linestyle="--", label=f"{f} (train)")
    plt.plot(range(len(trees)), results[f]["test"], marker="o", label=f"{f} (test)")

plt.xticks(range(len(trees)), [str(t) for t in trees])
plt.xlabel("Number of Trees")
plt.ylabel("Test Accuracy")
plt.title("Effect of n_estimators and max_features on Random Forest Performance")
plt.legend()
plt.grid(True)
plt.show()
```
