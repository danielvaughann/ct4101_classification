```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
df_train = pd.read_csv("wildfires_training.csv")
df_test = pd.read_csv("wildfires_test.csv")
```

```python
#Convert categorical data to numeric
df_train["fire"] = np.where(df_train["fire"]=="no",0,1)
df_test["fire"] = np.where(df_test["fire"]=="no", 0, 1)
```

```python
#Test if yes and no data was converted to 1
df_train["fire"].unique()
```

```python
#Training label
Y = df_train["fire"]

#Training features
X = df_train[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]

#Training label
Y_test = df_test["fire"]

#Test features
X_test = df_test[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]
#Hyperparamter values
c_values = [0.001, 0.01, 0.1, 10, 50, 100]
penalty = ["l1", "l2", "elasticnet"]

results = {p: {"train": [], "test": []} for p in penalty}
        
def tuner(C, penalty):
    if penalty == "l1":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "liblinear", #allow logistic regression to run by updating solver which is needed for penalty parameter
)
    elif penalty == "l2":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "lbfgs",
)
    elif penalty == "elasticnet":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "saga",
            l1_ratio = 0.5,
)

    model.fit(X, Y)
    train_predict=model.predict(X)
    test_predict = model.predict(X_test)
    train_accuracy = accuracy_score(Y,train_predict)
    test_accuracy = accuracy_score(Y_test,test_predict)
    print(f"For c= {C} penalty={penalty} train_accuracy = {train_accuracy} accuracy={test_accuracy}")


    results[penalty]["train"].append(train_accuracy)
    results[penalty]["test"].append(test_accuracy)
    
for C in c_values:
    for p in penalty:
        tuner(C, p)


c_labels = ["0.001", "0.01", "0.1", "10", "50", "100"]
plt.figure(figsize=(8,5))
for p in penalty:
   # plt.plot(range(len(c_values)), results[p]["train"], marker="o", linestyle="--", label=f"{p} (train)")
   plt.plot(range(len(c_values)), results[p]["test"], marker="o", label=f"{p} (test)")
plt.xticks(range(len(c_values)), c_labels)
plt.xlabel("C value")
plt.ylabel("Test Accuracy")
plt.title("Effect of C and Penalty on Logistic Regression Performance")
plt.legend()
plt.grid(True)
plt.show()
```
