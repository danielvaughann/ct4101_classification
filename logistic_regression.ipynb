```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn import metrics
from sklearn.metrics import accuracy_score
```

```python
df_train = pd.read_csv("wildfires_training.csv")
df_test = pd.read_csv("wildfires_test.csv")
```

```python
#df_train.head()
```

```python
df_train["fire"] = np.where(df_train["fire"]=="no",0,1)
df_test["fire"] = np.where(df_test["fire"]=="no", 0, 1)
```

```python
df_train["fire"].unique()
```

```python
Y = df_train["fire"]

X = df_train[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]

Y_test = df_test["fire"]

X_test = df_test[["year","temp","humidity","rainfall","drought_code","buildup_index","day","month","wind_speed"]]
c_values = [0.001, 0.01, 0.1, 10, 50, 100]
penalty = ["l1", "l2", "elasticnet"]

def tuner(C, penalty):
    if penalty == "l1":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "liblinear",
)
    elif penalty == "l2":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "lbfgs",
)
    elif penalty == "elasticnet":
        model = LogisticRegression(
            C=C,
            penalty = penalty,
            solver = "saga",
            l1_ratio = 0.5,
)

    model.fit(X, Y)
    train_predict=model.predict(X)
    test_predict = model.predict(X_test)
    train_accuracy = accuracy_score(Y,train_predict)
    test_accuracy = accuracy_score(Y_test,test_predict)
    print(f"For c= {C} penalty={penalty} training accuracy={train_accuracy} accuracy={test_accuracy}")
for C in c_values:
    for p in penalty:
        tuner(C, p)
z
```
